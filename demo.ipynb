```python
# Import local modules
from nlp_parser import parse_user_request
from optimizer import optimize_gene
from assembler import assemble_plasmid
from main import choose_default_parts
```

```python
# Example input description
description = 'Express GFP in E. coli with IPTG induction'
gene_seq = 'ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG'
```

```python
# Parse the natural language description
reqs = parse_user_request(description)
design = choose_default_parts(reqs['organism'])
design.update(reqs)
design
```

```python
# Optimize the gene for the target organism
optimized_gene = optimize_gene(gene_seq, reqs['organism'])
optimized_gene
```

```python
# Assemble the final plasmid
record = assemble_plasmid(design, optimized_gene)
len(record.seq)
```
